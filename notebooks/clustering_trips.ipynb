```python
from sklearn.cluster import KMeans
import pandas as pd
```

```python
trips = pd.read_csv("../data/trips_small.csv")

# convert timestamp to weekday/weekend
datetime = pd.to_datetime(trips['time_start'], unit='s', utc=True)

# Monday 0, Sunday 6
def is_weekend(n):
    if n == 5 or n == 6:
        return True
    return False

is_weekend = datetime.dt.day_of_week.apply(is_weekend)
trips["is_weekend"] = is_weekend
trips.head()
```
